```python
def multiple(* num):
  s= sum(num)
  d= len(num)
  avg = s/d
  return s,d,avg
res_sum,res_len,res_avg=multiple(1,2,3,4,5,6,7,8,9)
print(f"{res_sum} {res_len} {res_avg}")
```
